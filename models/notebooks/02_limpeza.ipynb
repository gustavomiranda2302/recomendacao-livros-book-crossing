# 02 - Limpeza e transformação

**Objetivo:** tratar incongruências, nulos e registros inválidos (Preparação dos Dados).

**Lê:** `data/original/*.csv`  
**Gera:** `data/processed/{users,books,ratings}_clean.parquet`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

pd.set_option('display.max_columns', None)
original = Path("../data/original")

books = pd.read_csv(original / "Books.csv", low_memory=False)
ratings = pd.read_csv(original / "Ratings.csv")
users = pd.read_csv(original / "Users.csv")
```

# TRATAMENTOS APLICADOS



```python
#criando uma copia de cada tabela
users_original = users.copy()
books_original = books.copy()
ratings_original = ratings.copy()
```

```python
users['Age'] = users['Age'].apply(lambda x: x if 5 <= x <= 100 else np.nan)
```

No snippet acima, aplicamos um filtro para converter em nulas (NaN) as idades incongruentes (menores que 5 ou maiores que 100), mantendo o restante das linhas na tabela.

```python
books['Year-Of-Publication'] = pd.to_numeric(books['Year-Of-Publication'], errors='coerce')
```

Convertendo todos os dados da coluna Year of Publication pra numerico

```python
books.query("`Year-Of-Publication`  < 1800" )
```

```python

print(books["Year-Of-Publication"].median())
print(books["Year-Of-Publication"].mean())




```

Os cinco livros de exemplo vão de 1793 a 2017, uma diferença de 224 anos. Ou seja, o ano de lançamento varia muito de um livro para outro, e não existe um valor único que represente bem todos eles.

Se a gente preenchesse os anos 0 com a média (~1960) ou a mediana (1995), estaria colocando um valor quase certamente errado: Kabale und Liebe (1793), por exemplo, ficaria 202 anos fora do real.

Como os registros com ano 0 são só 1,7% do dataset, o mais coerente é removê-los e evitar inserir dados falsos.

```python

books['Year-Of-Publication'] = books['Year-Of-Publication'].apply(lambda x: x if 1800 <= x <= 2026 else np.nan)
```

aplicando um filtro pra permitir apenas livros publicados entre 1800 e 2026.

```python
location_split = users['Location'].str.split(',', n=2, expand=True)

users['City'] = location_split[0].str.strip()
users['State'] = location_split[1].str.strip()
users['Country'] = location_split[2].str.strip()

valores_invalidos = ['n/a', '', 'N/A', 'na', 'null', 'None']
users[['City', 'State', 'Country']] = users[['City', 'State', 'Country']].replace(valores_invalidos, np.nan)
```

A célula acima separa Location em City, State e Country, removendo espaços extras, e padroniza valores equivalentes a nulo (como n/a e string vazia) substituindo-os por NaN.

```python
books['Publisher'] = books['Publisher'].str.replace('&amp;', '&', regex=False)
books['Book-Title'] = books['Book-Title'].str.replace('&amp;', '&', regex=False)
```

O snippet acima substitui a entidade HTML '&amp' (resultado de um parsing incorreto do CSV original) pelo caractere '&' nas colunas Publisher e Book-Title, corrigindo a formatação do texto.

```python
books['ISBN'] = books['ISBN'].str.strip()
ratings['ISBN'] = ratings['ISBN'].str.strip()
```

acima tiramos os espaços em branco extras dos ISBN das tabelas books e ratings, garantindo que não ocorra nenhum problema durante o cruzamento das tabelas por diferenças de formatação

```python
ratings_zero = ratings[ratings['Book-Rating'] == 0]
ratings = ratings[ratings['Book-Rating'] != 0]
ratings = ratings[ratings['ISBN'].isin(books['ISBN'])]
```

A célula acima remove os ratings que são iguais a zero, que representam notas implicitas, mantendo apenas as explicitas (de 1 a 10) e remove ratings cujo ISBN não tem correspondente na tabela books

```python
print("Users:", users.shape)
print("Books:", books.shape)
print("Ratings:", ratings.shape, "(era", ratings_original.shape[0], "antes)")

print("\nNulos Age:", users['Age'].isnull().sum())
print("Nulos Year:", books['Year-Of-Publication'].isnull().sum())
print("Ratings implicitos: " , ratings_zero['Book-Rating'].count())
```

## Salvando os dados limpos

```python
PROC = Path("../data/clean")
PROC.mkdir(parents=True, exist_ok=True)

users.to_parquet(PROC / "users_clean.parquet")
books.to_parquet(PROC / "books_clean.parquet")
ratings.to_parquet(PROC / "ratings_clean.parquet")
```
