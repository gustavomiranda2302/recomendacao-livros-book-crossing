# 02 - Limpeza e transformação

**Objetivo:** tratar incongruências, nulos e registros inválidos (Preparação dos Dados).

**Lê:** `data/original/*.csv`  
**Gera:** `data/processed/{users,books,ratings}_clean.parquet`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

pd.set_option('display.max_columns', None)
original = Path("../data/original")

books = pd.read_csv(original / "Books.csv", low_memory=False)
ratings = pd.read_csv(original / "Ratings.csv")
users = pd.read_csv(original / "Users.csv")
```

# TRATAMENTOS APLICADOS



```python
#criando uma copia de cada tabela
users_original = users.copy()
books_original = books.copy()
ratings_original = ratings.copy()
```

```python
users['Age'] = users['Age'].apply(lambda x: x if 5 <= x <= 100 else np.nan)
```

No snippet acima, aplicamos um filtro para converter em nulas (NaN) as idades incongruentes (menores que 5 ou maiores que 100), mantendo o restante das linhas na tabela.

```python
books['Year-Of-Publication'] = pd.to_numeric(books['Year-Of-Publication'], errors='coerce')
```

Convertendo todos os dados da coluna Year of Publication pra numerico

```python
books.query("`Year-Of-Publication`  < 1800" )
```

```python

print(books["Year-Of-Publication"].median())
print(books["Year-Of-Publication"].mean())




```

Os cinco livros de exemplo vão de 1793 a 2017, uma diferença de 224 anos. Ou seja, o ano de lançamento varia muito de um livro para outro, e não existe um valor único que represente bem todos eles.

Se a gente preenchesse os anos 0 com a média (~1960) ou a mediana (1995), estaria colocando um valor quase certamente errado: Kabale und Liebe (1793), por exemplo, ficaria 202 anos fora do real.

Como os registros com ano 0 são só 1,7% do dataset, o mais coerente é removê-los e evitar inserir dados falsos.

```python
books['Year-Of-Publication'] = books['Year-Of-Publication'].apply(lambda x: x if 1800 <= x <= 2026 else np.nan)
```
